```python
# --- CELL 6: Build the results table ---

def build_results_table(prompts, test_cases, outputs, scores):
    rows = []
    for prompt_name in prompts:
        for tc in test_cases:
            key = f"{prompt_name}|{tc['test_id']}"
            output_text = outputs.get(key, "")
            score = scores.get(key, {})

            if not output_text:
                print(f"[MISSING] No output pasted yet for {key}")
            if key not in scores:
                print(f"[MISSING] No scores entered yet for {key}")

            rows.append({
                "test_id": tc["test_id"],
                "prompt_used": prompt_name,
                "output_summary": score.get("output_summary", ""),
                "accuracy": score.get("accuracy"),
                "helpfulness": score.get("helpfulness"),
                "format_adherence": score.get("format_adherence"),
                "completeness": score.get("completeness"),
                "evidence_grounding": score.get("evidence_grounding"),
                "safety_professionalism": score.get("safety_professionalism"),
                "clarity": score.get("clarity"),
                "failure_mode": score.get("failure_mode", ""),
                "revision_needed": score.get("revision_needed", ""),
                "full_output": output_text,
            })
    return pd.DataFrame(rows)

results_df = build_results_table(PROMPTS, TEST_CASES, OUTPUTS, SCORES)
```

CONCLUSION

The best-performing prompt from the library was P5_final_improved, which consistently produced the strongest outputs across the CGMP regulatory scenarios (T1, T3, T5, T7, and T9). This prompt performed better than the zero-shot and structured prompts because it provided additional context about the role, audience, tone, and communication requirements. By positioning the model as a “regulatory affairs advisor briefing a new quality manager with no CGMP background,” it guided the responses toward professional, accurate, and accessible explanations rather than generic summaries. The prompt also included important safeguards, such as avoiding alarmist language, clarifying that CGMP non-compliance does not automatically mean a drug is unsafe, and requiring the use of “not specified in source” when specific figures, timelines, or citations were unavailable. These constraints improved regulatory accuracy and reduced the risk of unsupported assumptions.

Compared with the P1_zero_shot prompts, P5_final_improved reduced several observed failure modes. It minimized overgeneralization of regulatory consequences by encouraging case-specific explanations of FDA actions such as seizures, injunctions, and Warning Letters. It reduced insufficient uncertainty handling by requiring explicit acknowledgment when information was not available rather than allowing the model to estimate penalties or timelines. It also reduced confusion between compliance findings and product safety risks by requiring clear statements that a CGMP observation represents a quality system deficiency and does not automatically confirm that a medication is unsafe. Compared with P2_structured prompts, it maintained the benefits of organized explanations while adding professional context and audience awareness.

In a real regulatory affairs or pharmaceutical quality workflow, this prompt could be used to generate training materials for new quality employees, draft internal compliance briefings, or support regulatory education initiatives. It would be particularly useful when translating complex FDA compliance concepts into explanations appropriate for non-specialists while maintaining scientific and regulatory precision.

However, some limitations remain. The prompt does not replace review by qualified regulatory professionals, especially for high-risk decisions involving product release, recalls, enforcement responses, or legal interpretation. The model may still require access to current FDA guidance and regulations because CGMP expectations and enforcement practices can evolve over time. Additionally, while the prompt improves consistency and safety, it cannot independently verify regulatory sources unless connected to appropriate reference materials. Therefore, the best use of P5_final_improved is as a regulatory communication and education support tool that complements, rather than replaces, expert judgment.
